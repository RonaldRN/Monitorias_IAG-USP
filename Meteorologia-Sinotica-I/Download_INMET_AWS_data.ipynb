
<font size=4 color='#002883'>**14. Fazer o download (ou coletar) dos dados do site do INMET de todo o seu histórico**

```python
# @title #### <font color='#002883'> Coletar o histórico de dados do período de interesse <font>
#%%capture

## COLETAR ARQUIVO ZIP DO INMET - Histórico 2007 a 2023
## "https://portal.inmet.gov.br/uploads/dadoshistoricos/2023.zip"
## Criando estrutura para captura dos dados de forma automatizada no lugar do ano será incluido a variável "i"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

for i in range(2005, 2024):
    # Mantendo /uploads/ na URL
    url = f"https://portal.inmet.gov.br/uploads/dadoshistoricos/{i}.zip"
    file = f"{i}.zip"

    try:
        print(f"Baixando {file}...")
        with requests.get(url, headers=headers, stream=True, timeout=60) as r:
            r.raise_for_status()
            # É obrigatório usar open() para gravar os bytes no disco
            with open(file, "wb") as f:
                for chunk in r.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
        print(f"{file} salvo com sucesso!")

    except requests.exceptions.RequestException as e:
        print(f"Falha ao baixar {file}: {e}")

    time.sleep(2)
```

<font size=4 color='#002883'> **23. Selecionar os arquivos do Período-1: 2007 até 2018**

```python
# @title #### <font color='#002883'> Selecionar o primeiro período de 2007 até 2018 <font>
#%%capture

#Criar df com os dados do ano

dados_2007_2018 = pd.DataFrame()

for ano in range(2007,2019):

    # Lista de arquivos
    with ZipFile(str(ano)+".zip", "r") as lista:
        #print(*lista.namelist(), sep="\n")
        listaArquivos = lista.namelist()

    # Separar os arquivos de SC
    listaSC = []
    for i in range(0,len(listaArquivos)):
        if re.findall("A806", listaArquivos[i]):
            listaSC.append(listaArquivos[i])
            #print(listaArquivos[i])

    # Selecionar o arquivo na listaSC e indicar a linha do "header" ==> cabeçalho
    for i in listaSC:
        with ZipFile(str(ano)+".zip") as file:
            with file.open(i) as f:
                df = pd.read_csv(f, sep=";", header=8, encoding="ISO-8859-1", decimal=",")

            #Meta dados inicio do CSV, transpor para transformar em coluna
            with ZipFile(str(ano)+".zip") as file:
                with file.open(i) as f:
                    #h = pd.read_csv(f, sep=";", nrows=7, encoding="ISO-8859-1",decimal=",").set_index("REGIÃO:").T
                    h = pd.read_csv(f, sep=";", nrows=7, encoding="ISO-8859-1",decimal=",")

                    # Incluir os meta dados no final df
                    for i in range(0,len(h.columns)):
                        df[h.columns[i].replace(":","")] = h.iloc[0,i]

                    # Concatenas os arquivos
                    dados_2007_2018 = pd.concat([df, dados_2007_2018], ignore_index=True)

#renomear colunas
dados_2007_2018 = dados_2007_2018.rename(columns={'DATA (YYYY-MM-DD)':'Data','HORA (UTC)':'Hora UTC','RADIACAO GLOBAL (KJ/m²)':'RADIACAO GLOBAL (Kj/m²)','ESTAÇÃO':'ESTACAO', 'DATA DE FUNDAÇÃO (YYYY-MM-DD)':'DATA DE FUNDACAO'})

dados_2007_2018
```

<font size=4 color='#002883'> **34. Graficando o diagrama de Gantt para a EMA Mirante A701**

```python
# @title #### <font color='#002883'> Criar gráfico de Gantt <font>
#%%capture

# Crear un gráfico de Gantt de disponibilidad de datos usando puramente Matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Variables a graficar (excluyendo columnas completamente vacías como Unnamed: 19)
columnas_validas = [col for col in A701_sorted.columns if col != 'Unnamed: 19']

fig, ax = plt.subplots(figsize=(12, 6), tight_layout=True)

# Dibujar barras para cada variable meteorológica
for idx, col in enumerate(columnas_validas):
    # Obtener booleanos de datos no nulos
    disponibilidad = A701_sorted[col].notna().astype(int)

    # Encontrar cambios de estado (0 a 1 o 1 a 0) para determinar los periodos continuos
    diff = disponibilidad.diff().fillna(0)

    # Identificar inicios y finales de bloques de datos
    inicios = disponibilidad.index[diff == 1].tolist()
    if disponibilidad.iloc[0] == 1:
        inicios.insert(0, disponibilidad.index[0])

    finales = disponibilidad.index[diff == -1].tolist()
    if disponibilidad.iloc[-1] == 1:
        finales.append(disponibilidad.index[-1])

    # Graficar cada bloque con barras horizontales
    for start, end in zip(inicios, finales):
        ax.barh(idx, (end - start).days, left=start, color='steelblue', height=0.6, align='center')

# Configurar los ejes y el estilo
ax.set_yticks(range(len(columnas_validas)))
ax.set_yticklabels(columnas_validas, fontsize=9)
ax.set_xlabel('Año', fontsize=10)
ax.set_title('Gráfico de Gantt - São Paulo (SP) - Mirante A701 (Disponibilidad de Datos)', fontsize=12, pad=15)
ax.grid(True, linestyle='--', alpha=0.3)

plt.show()

# Guardar el gráfico en Google Drive
path = '/content/drive/MyDrive/USP-CURSOS/MONITORIA/Meteorologia-Sinotica-I/Aula-2'
import os
os.makedirs(path, exist_ok=True)
fig.savefig(path + '/Grafico_Gantt_A701.png', format='png', dpi=300)
```

<font size=8 color='red'> **Climograma**

<font size=4 color='red'> O Climograma é um gráfico de representação das principais variáveis meteorológicas (precipitação e temperaturas) ao longo dos meses.

<font size=4 color='red'> Por convenção da OMM, um climograma é criado a partir de médias de um período padronizado em 30 anos. Mas neste caso, o climograma vai ser gerado com a data disponível.

<font size=4 color='red'> Para construir o nosso climograma, vamos mais uma vez usar o **groupby()** para agrupar os nossos dados de acordo com os meses **index.month** e, por fim, calcularmos as "normais" climatológicas.

<font size=4 color='#002883'>**47. Fazer o agrupamento por mêses e fazer o cálculo da média para obter o valor climático**

```python
# @title #### <font color='#002883'> Fazer climograma <font>
#%%capture

# FAZENDO UM CLIMOGRAMA

A701_climatologia = A701_media_diaria.groupby(A701_media_diaria.index.month).mean()
```

<font size=4 color='#002883'> **48. Para o caso da precipitação é necessário fazer um pre-processamento desses dados:**

<font size=4 color='#002883'> - Calcular os acumulados mensais para o período disponível.

<font size=4 color='#002883'> - Calcular a média dos acumulados e agregar esse novo valor ao DataFrame (sustituir o valor anterior)

```python
# @title #### <font color='#002883'> Calculando a precipitação acumulada mensal <font>
#%%capture

# Calculando acumulados de precipitação mensais
# Fazer médias mensais dos acumulados de precipitação

# Acumulado mensal
A701_pp_mensal = A701_media_diaria['PRECIPITAÇÃO TOTAL, HORÁRIO (mm)'].groupby(pd.Grouper(freq = '1M')).sum()

# Média climatológica de precipitação
A701_climatologia['PRECIPITAÇÃO TOTAL, HORÁRIO (mm)'] = A701_pp_mensal.groupby(A701_pp_mensal.index.month).mean()
```

<font size=4 color='#002883'> **49. Os climogramas apresentam dois eixos Y. O primeiro eixo é de precipitação e o segundo de temperatura, o que significa que todas as variáveis compartilham o mesmo eixo x dos meses.**

<font size=4 color='#002883'> **O Matplotlib e Proplot podem criar eixos "gêmeos", tanto para o eixo x .twinx() quanto y .twiny():**

```python
# @title #### <font color='#002883'> Figura do climograma <font>
#%%capture

# Climograma usando apenas Matplotlib
fig, ax = plt.subplots(figsize=(10, 4), tight_layout=True)

# Eixos gêmeos em X para as temperaturas
ax2 = ax.twinx()

meses = list(range(1, 13))
meses_labels = ['JAN', 'FEV', 'MAR', 'ABR', 'MAI', 'JUN', 'JUL', 'AGO', 'SET', 'OUT', 'NOV', 'DEZ']

# Precipitação (Eixo Y Esquerdo)
ax.bar(meses, A701_climatologia['PRECIPITAÇÃO TOTAL, HORÁRIO (mm)'], color='blue', alpha=0.5, label='Precipitação')
ax.set_ylabel('Precipitação (mm/mês)')
ax.set_xlabel('Meses')
ax.set_xticks(meses)
ax.set_xticklabels(meses_labels)

# Temperaturas (Eixo Y Direito)
line_ins = ax2.plot(meses, A701_climatologia['TEMPERATURA DO AR - BULBO SECO, HORARIA (°C)'], color='grey', label='INS')
line_max = ax2.plot(meses, A701_climatologia['TEMPERATURA MÁXIMA NA HORA ANT. (AUT) (°C)'], color='red', label='MAX')
line_min = ax2.plot(meses, A701_climatologia['TEMPERATURA MÍNIMA NA HORA ANT. (AUT) (°C)'], color='blue', label='MIN')
ax2.set_ylabel('Temperatura (°C)')

ax.set_title('Climograma - São Paulo - Mirante - A701')
ax.grid(True, linestyle='--', alpha=0.3)

# Unificando a legenda em um único quadro
lines = line_ins + line_max + line_min
labels = [l.get_label() for l in lines]
ax2.legend(lines, labels, loc='upper right', frameon=False)

plt.show()

# Salvando o plot
path = '/content/drive/MyDrive/USP-CURSOS/MONITORIA/Meteorologia-Sinotica-I/Aula-2'
fig.savefig(path + '/Climograma-A701.png', format='png', dpi=300)
```
